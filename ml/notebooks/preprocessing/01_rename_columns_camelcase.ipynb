# Renombrar columnas a camelCase

Este notebook estandariza los nombres de las columnas del dataset
al formato camelCase para mantener consistencia en los procesos
de limpieza y modelado de datos.


```python
# Importando Librerías
import pandas as pd
import re
```

```python
# cargar el dataset
df = pd.read_csv("../../data/processed/gym_churn_cleaned_2.csv")
df.head()
```

```python
# Función CamelCase
def a_camel_case(columna):
    # Si ya es camelCase, la dejamos
    if columna[0].islower() and any(c.isupper() for c in columna[1:]):
        return columna

    # Reemplazar separadores por espacios
    columna = re.sub(r'[_\-]+', ' ', columna)
    columna = re.sub(r'[^a-zA-Z0-9 ]', '', columna)

    partes = columna.lower().split()
    return partes[0] + ''.join(p.capitalize() for p in partes[1:])
```

```python
ruta_salida = "../../data/processed/gym_churn_camelcase.csv"

df.to_csv(ruta_salida, index=False)

print(f"Dataset guardado correctamente en: {ruta_salida}")
```
